# ACT - American College Testing

## 2012

```python
url = 'https://nces.ed.gov/programs/digest/d16/tables/dt16_226.60.asp'
tables = pd.read_html(url, header=0)
df = tables[3]
```

```python
columns_2012 = [col for col in df.columns if '2012' in str(col)]
columns_2012.insert(0, 'State')
df_2012 = df[columns_2012]
df_2012 = df_2012.dropna().reset_index(drop=True)
```

```python
df_2012.index = ['df_2012'] + list(range(1, len(df_2012)))

df = df_2012

df = df.drop(index=1)

df.columns = df.iloc[0]
df = df[1:]
df.reset_index(drop=True, inplace=True)

df.set_index(df.columns[0], inplace=True)

for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df_2012 = df

print(df_2012.head())
```

# SAT

2012-2019

## 2012

```python
url = "https://nces.ed.gov/programs/digest/d14/tables/dt14_226.40.asp"

tables = pd.read_html(url, header=0)

df = tables[3]

thistitle = 'sat_2012'
```

```python
df['State'] = df['State'].astype(str).str.rstrip('34')

df = df[df['State'].notna() & (df['State'].str.strip() != '')]

rows = df['State'].str.strip().isin(['District of', 'Colombia'])
if rows.sum() == 2:
    combined_row = df.loc[rows].drop(columns=['State']).apply(pd.to_numeric, errors='coerce').sum()
    df = df[~rows] 
    df.loc[len(df)] = ['District of Columbia'] + combined_row.tolist()

df = df.iloc[:, :-2]
df = df.loc[:, df.columns.str.startswith('2012') | df.columns.str.startswith('State')]
df.index = [thistitle if i == df.index[0] else i for i in df.index]

df.iloc[0] = ["State", "Reading", "Mathematics", "Writing"]
df.columns = df.iloc[0]
df = df.iloc[1:].reset_index(drop=True)

df = df.iloc[1:].reset_index(drop=True)

df.set_index(df.columns[0], inplace=True)

df = df[df.index != 'nan']

df = df.drop("District of", errors='ignore')
df.rename(index={"Columbia": "District of Columbia"}, inplace=True)
```

```python
sat_2012 = df

print(sat_2012)
```

## 2017

```python
url = "https://nces.ed.gov/programs/digest/d20/tables/dt20_226.40.asp"

tables = pd.read_html(url, header=0)

df_2017 = tables[3] 

df= df_2017.loc[:, df_2017.columns.str.startswith('2017')  | df_2017.columns.str.startswith('State')]

df= df.iloc[1:].reset_index(drop=True)
df.set_index(df.columns[0], inplace=True)

df= df.iloc[1:]

df.index = ['SAT_2017' if str(i).strip() in ['1', 'State'] else i for i in df.index]

df= df[~df.iloc[:, 0].astype(str).str.strip().eq("2")]
df= df.reset_index()

df= df.loc[:, df.iloc[0] != "SD"]
df.set_index(df.columns[0], inplace=True)

df= df.iloc[1:]
df= df.iloc[:, :-1]

df.columns = ['Total', 'ERW', 'Math']
```

```python
sat_2017 = df

sat_2017.index.name = 'SAT_2017'
print (sat_2017)
```
